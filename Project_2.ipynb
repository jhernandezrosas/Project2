```python
url = '/content/drive/MyDrive/Project 2/TrendGenerator-Table-q24-05-05-26 12-47.csv'
data_states = pd.read_csv(url, sep=',', engine='python', skiprows=[0,1], header=0)
data_states.head()
```

```python
data_states['Total'] = pd.to_numeric(
    data_states['Total'].astype(str).str.replace(',', '', regex=False),
    errors='coerce'
)
data_states['Total'] = data_states['Total'].fillna(0)
```

**Table 1 - Interactive Bar Chart**

```python
import altair as alt
alt.data_transformers.enable("vegafusion")

# Define the custom years for the selection
custom_years = [2000, 2005, 2010, 2015, 2020, 2024]

# Create a slider for the year, allowing selection of all years
year_selector = alt.binding_range(min=min(custom_years),
                                max=max(custom_years),
                                step=5,
                                name='Year:')

select_year = alt.selection_point(name='year_selection', fields=['YEAR'], bind=year_selector, value=2000)

# Create a dropdown menu for Degree Description
description_dropdown = alt.binding_select(options=merged_data_with_descriptions['Description'].unique().tolist(),
                                         name='Degree Description:')
select_description = alt.selection_point(fields=['Description'], bind=description_dropdown)

# Create the bar chart
chart = alt.Chart(merged_data_with_descriptions).mark_bar().encode(
    x=alt.X('Description:N', sort='-y', title='Description'),
    y=alt.Y('Total Count:Q', title='Total Count'),
    tooltip=['YEAR', 'Description', 'Total Count']
).add_params(
    select_year,
    select_description
).transform_filter(
    select_year
).transform_filter(
    select_description
).transform_filter(
    alt.datum['Total Count'] > 200 # Filter for Total Count > 200
).properties(
    title='Total Count by Description for Selected Year (Count > 200)'
).interactive()

chart

!pip install "vl-convert-python>=1.6.0"
```

**Table 2**

```python
import altair as alt
import pandas as pd

# 1. Aggregate data to count degrees per CIPCODE and Year
#    'merged_data_filtered' contains individual entries for each degree awarded with AWLEVEL == 5
degrees_per_year_cipcode = merged_data_filtered.groupby(['YEAR', 'CIPCODE']).size().reset_index(name='Degrees_Awarded')

# 2. Merge with descriptions to get the 'Description' for each CIPCODE
#    'cip_descriptions_df' is available from previous steps
degrees_growth_data = pd.merge(degrees_per_year_cipcode, cip_descriptions_df, on='CIPCODE', how='left')

# Drop rows where Description might be missing if a CIPCODE from merged_data_filtered wasn't in cip_descriptions_df
degrees_growth_data.dropna(subset=['Description'], inplace=True)

# Calculate total degrees awarded for each description across all years to identify top descriptions
total_degrees_by_description = degrees_growth_data.groupby('Description')['Degrees_Awarded'].sum().reset_index(name='Overall_Total_Degrees')

# Get the top N descriptions based on their overall total degrees
top_n = 15 # You can adjust this number to show more or fewer lines
top_descriptions = total_degrees_by_description.nlargest(top_n, 'Overall_Total_Degrees')['Description'].tolist()

# Filter the data to include only the top descriptions
filtered_degrees_growth_data = degrees_growth_data[degrees_growth_data['Description'].isin(top_descriptions)]

# Create a selection for the descriptions (to act like buttons), allowing multiple selections
legend_selection = alt.selection_point(fields=['Description'], bind='legend', name='description_selection', toggle=True)

# Add an interval selection for the YEAR axis
year_interval_selection = alt.selection_interval(encodings=['x'], name='year_interval')

# Create the line chart
line_chart = alt.Chart(filtered_degrees_growth_data).mark_line(point=True).encode(
    x=alt.X('YEAR:O', axis=alt.Axis(format='d'), title='Year'), # 'O' for ordinal to treat years as discrete categories
    y=alt.Y('Degrees_Awarded:Q', title='Degrees Awarded'),
    color=alt.Color('Description:N', legend=alt.Legend(title="Degree Description")),
    tooltip=['YEAR:O', 'Description:N', 'Degrees_Awarded:Q'],
    opacity=alt.condition(legend_selection, alt.value(1), alt.value(0.2)) # Highlight selected descriptions
).add_params(
    legend_selection,
    year_interval_selection
).transform_filter(
    year_interval_selection # Filter based on the year interval selection
).properties(
    title=f'Growth of Top {top_n} Degrees Over Time (Select Year Range)'
).interactive() # Allows zooming and panning

line_chart
```

```python
import pandas as pd
import altair as alt

# Calculate growth for each description
growth_data = []

for desc in degrees_growth_data['Description'].unique():
    subset = degrees_growth_data[degrees_growth_data['Description'] == desc].sort_values('YEAR')

    if len(subset) > 1:
        first_year_data = subset.iloc[0]
        last_year_data = subset.iloc[-1]

        initial_degrees = first_year_data['Degrees_Awarded']
        final_degrees = last_year_data['Degrees_Awarded']
        initial_year = first_year_data['YEAR']
        final_year = last_year_data['YEAR']

        # Avoid division by zero for growth calculation
        if initial_degrees > 0:
            growth_percent = ((final_degrees - initial_degrees) / initial_degrees) * 100
            growth_data.append({
                'Description': desc,
                'Initial_Degrees': initial_degrees,
                'Final_Degrees': final_degrees,
                'Initial_Year': initial_year,
                'Final_Year': final_year,
                'Growth_Percent': growth_percent,
                'Years_Span': final_year - initial_year
            })
        elif final_degrees > 0: # If started at 0 but grew
            growth_data.append({
                'Description': desc,
                'Initial_Degrees': initial_degrees,
                'Final_Degrees': final_degrees,
                'Initial_Year': initial_year,
                'Final_Year': final_year,
                'Growth_Percent': float('inf'), # Represent as infinite growth
                'Years_Span': final_year - initial_year
            })

growth_df = pd.DataFrame(growth_data)

# Sort by growth percent to find the fastest growing
growth_df_sorted = growth_df.sort_values(by='Growth_Percent', ascending=False)

# Display the top 10 fastest growing fields
print("Top 10 Fastest Growing Fields:")
display(growth_df_sorted.head(10))

# Identify the fastest growing field
if not growth_df_sorted.empty:
    fastest_growing_field = growth_df_sorted.iloc[0]
    print(f"\nThe fastest growing field is: {fastest_growing_field['Description']} "
          f"with a growth of {fastest_growing_field['Growth_Percent']:.2f}% "
          f"(from {fastest_growing_field['Initial_Degrees']} degrees in {fastest_growing_field['Initial_Year']} "
          f"to {fastest_growing_field['Final_Degrees']} degrees in {fastest_growing_field['Final_Year']}).")
else:
    print("No growth data available.")


# Visualize the growth rates
# Filter out infinite growth for a meaningful bar chart, or handle it specially
chart_data = growth_df_sorted[growth_df_sorted['Growth_Percent'] != float('inf')].head(20)

bar_chart = alt.Chart(chart_data).mark_bar().encode(
    x=alt.X('Growth_Percent:Q', title='Percentage Growth (from initial to final year)'),
    y=alt.Y('Description:N', sort='-x', title='Field Description'),
    tooltip=[
        'Description:N',
        'Initial_Degrees:Q',
        'Final_Degrees:Q',
        'Initial_Year:O',
        'Final_Year:O',
        alt.Tooltip('Growth_Percent:Q', format='.2f', title='Growth %')
    ]
).properties(
    title='Top 20 Fields by Percentage Growth (from first to last recorded year)'
).interactive()

bar_chart
```
